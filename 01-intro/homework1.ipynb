```python
# Q1

df = pd.read_csv('car_fuel_efficiency_2026.csv')
df.head()
```

```python
# Q2

print(df.shape)
print(f'Number of records in dataset: {df.shape[0]}')
```

```python
# Q3

print(f'Number of fuel types presented in the dataset: {df['fuel_type'].nunique()}')
```

```python
# Q4

print('Missing values by column:')
print(df.isnull().sum())

print(f'\nNumber of columns with missing values: {df.isnull().any().sum()}')
```

```python
# Q5

print(df['origin'].unique())
print(f"Maximum fuel efficiency (MPG) of cars from Asia: {df[df['origin'] == 'Asia']['fuel_efficiency_mpg'].max()}")
# df.loc[df['origin'] == 'Asia', 'fuel_efficiency_mpg'].max()
```

```python
# Q6

# Median value of `horsepower` column
median_horsepower = df['horsepower'].median()

# Most frequent value of `horsepower` column
mode_horsepower = df['horsepower'].mode()[0]

# Fill missing values in `horsepower` column with the most frequent value
df['horsepower'] = df['horsepower'].fillna(mode_horsepower)

# New median
median_horsepower_after = df['horsepower'].median()

print(f'Median horsepower before filling with most frequent value: {median_horsepower}')
print(f'After filling: {median_horsepower_after}')
```

```python
# Q7

# First 7 values of the columns `vehicle_weight` and `model_year` for cars from Asia
X_df = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7)
display(X_df)

# Convert it to NumPy array
X = X_df.to_numpy()
print(X)

XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inv.dot(X.T).dot(y)
print(w)

print(f'Sum of the elements in the resulting weight vector: {w.sum()}')
```
